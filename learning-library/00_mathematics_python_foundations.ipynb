## Lesson 1 — Arithmetic, units and variables

A variable is a name representing a value. Python assignment stores a value; mathematical equality expresses a relationship. Python uses +, -, *, / and ** for arithmetic. Parentheses control order.

For constant power, $E=Pt$. A 6 MW battery delivering rated power for 3 hours supplies $6\times3=18$ MWh, neglecting losses.

Conversions: $1\,\mathrm{kW}=1000\,\mathrm{W}$ and $1\,\mathrm{MW}=1000\,\mathrm{kW}$.

**Python translation:** assign input values, multiply the two variables with *, store the result, and display it with print(). A variable name ending in _mw or _h helps identify its units.

```python
power_mw = 6
duration_h = 3
energy_mwh = power_mw * duration_h
print("Energy (MWh):", energy_mwh)
```

## Lesson 2 — Linear equations and rearranging formulas

An equation stays balanced when the same operation is applied to both sides. To solve $4x+5=21$, subtract 5 from both sides: $4x=16$. Divide by 4: $x=4$. Verify by substitution: $4(4)+5=21$.

Ohm's law: $V=IR$. Divide by $I$ (assuming $I\ne0$) to obtain $R=V/I$. For 120 V and 4 A, resistance is $30\,\Omega$.

**Python translation:** import sympy as sp gives the library a short name. sp.symbols creates an unknown; sp.Eq(left,right) represents equality; sp.solve(equation,x) finds x. / performs division.

```python
import sympy as sp
x = sp.symbols("x")
equation = sp.Eq(4*x + 5, 21)
print(sp.solve(equation, x))
voltage_v = 120
current_a = 4
resistance_ohm = voltage_v / current_a
print(resistance_ohm)
```

## Lesson 3 — Polynomials, factoring and quadratic equations

The identity $a^2-b^2=(a-b)(a+b)$ allows factoring. Thus $(x^2-16)/(x-4)=x+4$ provided $x\ne4$. Cancelling does not remove an original domain restriction.

To solve $x^2-25=0$, factor $(x-5)(x+5)=0$. A product is zero if at least one factor is zero; hence $x=5$ or $x=-5$. Check both by substitution.

**Python translation:** ** represents powers, not ^. sp.factor rewrites a polynomial as factors; sp.cancel simplifies rational expressions; sp.solve finds roots. Keep original domain restrictions separately.

```python
import sympy as sp
x = sp.symbols("x")
print(sp.factor(x**2 - 16))
print(sp.cancel((x**2 - 16)/(x - 4)))
print(sp.solve(sp.Eq(x**2 - 25, 0), x))
```

## Lesson 4 — Trigonometry and radians

For a right triangle, $\sin\theta=\mathrm{opposite}/\mathrm{hypotenuse}$ and $\cos\theta=\mathrm{adjacent}/\mathrm{hypotenuse}$. Convert degrees to radians using $\theta_{\mathrm{rad}}=\theta_{\mathrm{deg}}\pi/180$.

At 60°, $\theta=\pi/3$ and $\cos\theta=1/2$. A length of 10 m inclined at 60° to horizontal has a horizontal projection of $10\cos60^\circ=5$ m.

**Python translation:** import math loads the standard mathematical library; math.radians converts degrees; math.cos expects radians; sp.cos(sp.pi/3) gives an exact symbolic value.

```python
import math
import sympy as sp
angle_deg = 60
angle_rad = math.radians(angle_deg)
projection_m = 10 * math.cos(angle_rad)
print(angle_rad, projection_m)
print(sp.cos(sp.pi/3))
```

## Lesson 5 — Exponents, logarithms and functions

For positive integer exponents, $3^4=3\cdot3\cdot3\cdot3=81$. To solve $3^x=81$, use the same base: $3^x=3^4$, so $x=4$. For $b>0$, $b\ne1$ and $y>0$, $b^x=y$ implies $x=\log_b(y)$.

A function maps an input to an output. If $f(t)=6t$ describes energy at constant 6 MW, then $f(3)=18$ MWh.

```python
import math
print(3**4)
print(math.log(81, 3))
def energy_mwh(time_h):
    return 6 * time_h
print(energy_mwh(3))
```

## Lesson 6 — Derivatives

A derivative is the instantaneous rate of change of a function. For positive integer $n$, $\frac{d}{dx}x^n=nx^{n-1}$. The derivative of a constant is zero. Differentiate each term separately.

For $f(x)=2x^3-5x+7$, $f'(x)=6x^2-5$. At $x=2$, the slope is $19$. In energy engineering, $dE/dt=P$ when units are consistent (MWh per hour is MW).

**Python translation:** sp.diff(expression, variable) differentiates symbolically; .subs(x,2) substitutes 2 for x.

```python
import sympy as sp
x = sp.symbols("x")
f = 2*x**3 - 5*x + 7
df = sp.diff(f, x)
print(df)
print(df.subs(x, 2))
```

## Lesson 7 — Integrals

A definite integral accumulates a quantity over an interval. For positive integer $n$, an antiderivative of $x^n$ is $x^{n+1}/(n+1)+C$. Evaluate at the upper limit and subtract the value at the lower limit.

For synthetic power $P(t)=3t$ MW, with t in hours and coefficient 3 MW/h, over $0\le t\le4$:
$$E=\int_0^4 3t\,dt=[1.5t^2]_0^4=24\ \mathrm{MWh}.$$

**Python translation:** sp.integrate(expression, (variable, lower, upper)) evaluates a definite integral. The tuple specifies the integration variable and limits.

```python
import sympy as sp
t = sp.symbols("t")
P = 3*t
E = sp.integrate(P, (t, 0, 4))
print(E)
```

## Lesson 8 — Systems of equations

Multiple equations can constrain several unknowns. Solve $a+b=12$ and $2a-b=3$. Adding eliminates b: $3a=15$, so $a=5$. Substitute into the first equation: $b=7$. Verify both original equations.

**Python translation:** sp.symbols creates two unknowns. Create each equation with sp.Eq. Pass lists of equations and variables to sp.solve; dict=True returns solutions as a list of dictionaries.

```python
import sympy as sp
a, b = sp.symbols("a b")
eq1 = sp.Eq(a+b, 12)
eq2 = sp.Eq(2*a-b, 3)
print(sp.solve([eq1,eq2], [a,b], dict=True))
```
